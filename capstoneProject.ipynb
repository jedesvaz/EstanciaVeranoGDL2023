```python
#select predictor variables and target variable as separate data sets  (ALL PREDICTORS)
predvar1= data[['CONCENTR_VD', 'EDAD', 'Cont_min_oseo', 'Angulo_fase', 'Masa_magra',
       'IMC']]

target = data.RECURRENCIA
```

```python
#select predictor variables and target variable as separate data sets  (AGE, VITAMIN D CONCENTRATION AND BODY PHASE ANGLE)
predvar2= data[['CONCENTR_VD', 'EDAD', 'Angulo_fase',]]
```

```python
#select predictor variables and target variable as separate data sets  (BONE MINERAL DENSITY, VITAMIN D CONCENTRATION, LEAN BODY MASS AND BODY PHASE ANGLE)
predvar3= data[['CONCENTR_VD', 'Cont_min_oseo', 'Angulo_fase', 'Masa_magra']]
```

```python
# split data into train and test sets
predictors=predvar1.copy()
pred_train, pred_test, tar_train, tar_test = train_test_split(predictors, target, test_size=.2, random_state=123)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import sklearn.metrics
```

```python
#prepocessing data
ss = StandardScaler()
pred_train = ss.fit_transform(pred_train)
```

```python
#fitting data
# Use the fit method of LogisticRegression to train model parameters.
lr = LogisticRegression()
lr.fit(pred_train, tar_train)
```

```python
#predicting data
#testX = [[200,8]]
pred_test = ss.transform(pred_test)

print("Value to be predicted: ",pred_test)
label = lr.predict(pred_test)
print("predicted label = ", label)

# Output the predicted probability.
prob = lr.predict_proba(pred_test)
print("probability = ",prob)
```

```python
#Model with all predictors
sklearn.metrics.confusion_matrix(tar_test,label)
```

```python
#Model with all predictors
sklearn.metrics.accuracy_score(tar_test, label)
```

```python
#Model with AGE, VITAMIN D CONCENTRATION AND BODY PHASE ANGLE
sklearn.metrics.accuracy_score(tar_test, label)
```
